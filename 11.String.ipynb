```python
####  count characters ###

text = input("Enter a string: ")

count = len(text)

print("Number of characters:", count)
```

```python
### count perticular character ###

text = input("Enter a string: ")

count = text.count("e")

print("Number of a:", count)
```

```python
## count vowels ###

text = input("Enter a string: ")

count = 0

for char in text:
    if char in "aeiou":
        count += 1

print("Number of vowels:", count)
```
