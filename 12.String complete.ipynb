```python
### STRING IMMUTABILITY ###


name = "Python"

name[0] = "J"

print(name)
```

```python
name = "Python"

name = "J" + name[1:]

print(name)
```

```python
text = "I like Java"

text = text.replace("Java", "Python")

print(text)
```

```python
city = "Chennai"

city = "Bangalore"

print(city)
```

```python
### STRING CONCATENATION ###

first_name = "Tamil"
last_name = "maran"

full_name = first_name + last_name

print(full_name)
```

```python
### MEMBERSHIP OPERATORS ###

## IN ##

text = "Python"

print("Py" in text)
print("Java" in text)
```

```python
email = "user@gmail.com"

if "@" in email:
    print("Valid email format")
```

```python
## NOT IN ##

text = "Python Programming"

if "Java" not in text:
    print("Java is not present")
```

```python
## F STRING ##

name = "Tamilmaran"

print(f"My name is {name}")
```
