```python
marks = []

for i in range(5):
    mark = int(input("Enter mark: "))
    marks.append(mark)

result = any(mark >= 90 for mark in marks)

if result:
    print("At least one student scored 90 or above")
else:
    print("Nobody scored 90 or above")
```

```python
### LIST COMPREHENSION ##

numbers = [1, 2, 3, 4, 5]

squares = []

for num in numbers:
    squares.append(num * num)

print(squares)
```

```python
numbers = [1, 2, 3, 4, 5]

squares = [num * num for num in numbers]

print(squares)
```

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

even_numbers = [num for num in numbers if num % 2 == 0]

print(even_numbers)
```
